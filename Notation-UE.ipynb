```python
from tkinter import *
from tkinter import ttk
from PIL import Image, ImageDraw, ImageTk

fenetre = Tk()
fenetre.title("Moyenne BUT 1 - Semestre 1")

frm = ttk.Frame(fenetre)
frm.grid()

noms = ["R101", "R102", "R103", "R104", "R105", "R106", "R107",
        "R108", "R109", "R110", "R111", "R112", "R113", "R114", "R115",
        "SAE11", "SAE12", "SAE13", "SAE14", "SAE15", "SAE16"]

ue1 = [10, 10, 7, 7, 0, 5, 0, 6, 0, 5, 4, 2, 5, 5, 0, 20, 20, 0, 0, 0, 7]
ue2 = [4, 0, 2, 8, 6, 0, 0, 0, 0, 5, 5, 2, 9, 9, 3, 0, 0, 29, 0, 0, 7]
ue3 = [4, 0, 2, 0, 0, 5, 15, 6, 4, 5, 5, 2, 0, 0, 3, 0, 0, 0, 20, 20, 7]
coefs = [ue1, ue2, ue3]

cases = []
for i in range(len(noms)):
    texte = ttk.Label(frm, text=noms[i])
    texte.grid(column=0, row=i)
    saisie = Entry(frm, width=8)
    saisie.grid(column=1, row=i)
    cases.append(saisie)


def dessiner(moyennes):
    image = Image.new("RGB", (300, 250), "white")
    dessin = ImageDraw.Draw(image)

    dessin.line((20, 220, 280, 220), fill="black")

    for u in range(3):
        moyenne = moyennes[u]

        if moyenne < 8:
            couleur = "red"
        elif moyenne < 10:
            couleur = "orange"
        else:
            couleur = "green"

        hauteur = moyenne / 20 * 180
        x = 40 + u * 90

        dessin.rectangle((x, 220 - hauteur, x + 60, 220), fill=couleur)
        dessin.text((x + 15, 220 - hauteur - 15), str(moyenne), fill="black")
        dessin.text((x + 20, 228), "UE" + str(u + 1), fill="black")

    photo = ImageTk.PhotoImage(image)
    graphique.config(image=photo)
    graphique.image = photo  


def calculer():
    resultat_texte = ""
    moyennes = []
    for u in range(3):
        somme = 0
        total = 0
        for i in range(len(noms)):
            note = cases[i].get()
            if note != "":  
                somme = somme + float(note.replace(",", ".")) * coefs[u][i]
                total = total + coefs[u][i]
        if total > 0:
            moyenne = round(somme / total, 2)
        else:
            moyenne = 0
        moyennes.append(moyenne)
        resultat_texte = resultat_texte + "UE" + str(u + 1) + " : " + str(moyenne) + "\n"
    resultat.config(text=resultat_texte)
    dessiner(moyennes)


bouton = Button(frm, text="Calculer", command=calculer)
bouton.grid(column=0, row=len(noms))

resultat = ttk.Label(frm, text="")
resultat.grid(column=0, row=len(noms) + 1)

graphique = ttk.Label(frm)
graphique.grid(column=2, row=0, rowspan=10)

fenetre.mainloop()
```
