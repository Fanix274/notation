```python
som_1=0
coe_1=0
moy_1=0

som_2=0
coe_2=0
moy_2=0

som_3=0
coe_3=0
moy_3=0

for i in range (101 , 116):
    note=float(input("Donnez moi la note de la ressources R{}".format(i)))
    coeff1=float(input("Donnez moi le coefficient de la ressources R{} pour l'UE 1".format(i)))
    coeff2=float(input("Donnez moi le coefficient de la ressources R{} pour l'UE 2".format(i)))
    coeff3=float(input("Donnez moi le coefficient de la ressources R{} pour l'UE 3".format(i)))
    som_1+=note*coeff1
    som_2+=note*coeff2
    som_3+=note*coeff3
    coe_1+=coeff1
    coe_2+=coeff2
    coe_3+=coeff3
for i in range (11 , 13):
    note=float(input("Donnez moi la note de la SAE SAE{}".format(i)))
    coeff1=float(input("Donnez moi le coefficient de la SAE SAE{} pour l'UE 1".format(i)))
    coeff2=float(input("Donnez moi le coefficient de la SAE SAE{} pour l'UE 2".format(i)))
    coeff3=float(input("Donnez moi le coefficient de la SAE SAE{} pour l'UE 3".format(i)))
    som_1+=note*coeff1
    som_2+=note*coeff2
    som_3+=note*coeff3
    coe_1+=coeff1
    coe_2+=coeff2
    coe_3+=coeff3
moy_1=som_1/coe_1
moy_2=som_2/coe_2
moy_3=som_3/coe_3
print(f"{moy_1:.2f}")
print(f"{moy_2:.2f}")
print(f"{moy_3:.2f}")
```
